```python
clean_df = raw_df.copy()
```

```python
missing_summary = pd.DataFrame({
    "Missing Count": clean_df.isnull().sum(),
    "Missing Percentage": (clean_df.isnull().sum() / len(clean_df) * 100).round(2)
})

missing_summary = missing_summary.sort_values(
    by="Missing Percentage",
    ascending=False
)

display(missing_summary)
```

```python
print(" Printing the unique values of every column :")
for col in clean_df.columns[1:]:
    print(f" The unique values in {col} are : {clean_df[col].unique()}")
```

```python
# Count duplicate rows

duplicate_count = clean_df.duplicated().sum()

print(f"Total Duplicate Rows : {duplicate_count}")
```

```python
if duplicate_count > 0:
    duplicate_rows = clean_df[clean_df.duplicated(keep=False)]
    display(duplicate_rows.sort_values(by=duplicate_rows.columns.tolist()))
else:
    print("No duplicate rows found.")
```

**As i can see here most of the duplicate values have a weird pattern that is they are appearing just after another also most of them are appearing on the same date which indicates that there is a very high probability that those values are the real duplicate values and hence i am dropping those values**

```python
from encodings.punycode import T

clean_df.drop_duplicates(inplace=True)
```

```python
clean_df.head()
```

```python
clean_df.drop(columns=["Timestamp"], inplace = True)
clean_df = clean_df[clean_df["DAILY_STRESS"] != '01-01-2000']
clean_df['AGE'] = clean_df['AGE'].map({ "Less than 20": 0, "21 to 35": 1, "36 to 50": 2, "51 or more": 3 })
clean_df = pd.get_dummies(clean_df, columns=['GENDER'], drop_first=True)
clean_df['DAILY_STRESS'] = clean_df['DAILY_STRESS'].astype("int64")
```

```python
clean_df.to_csv('dataset\\clean_dataset.csv', index=False)
```
